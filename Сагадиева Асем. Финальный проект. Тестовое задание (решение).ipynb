```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import statsmodels.api as sm
```

```python
df = pd.read_excel('Данные для тестового задания.xlsx', sheet_name='Данные АБ тестов')
df.head()
```

```python
# 8 задание

for exp in df['experiment_num'].unique():
  c = df[(df['experiment_num'] == exp) & (df['experiment_group'] == 'control')][
      'revenue'
  ]
  t = df[(df['experiment_num'] == exp) & (df['experiment_group'] == 'test')][
      'revenue'
  ]
  if len(t) > 0:
    stat, pval = stats.ttest_ind(c, t, equal_var=False)
    print(f'Эксперимент {exp}: p-value = {pval:.4f}')
  else:
    print(f'Эксперимент {exp}: нет тестовой группы')
```
